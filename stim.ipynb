```python
d = 3
x_stabilizers = []
x_stab_coords = []
z_stabilizers = []
z_stab_coords = []
for row in range(d):
    for col in range(d):
        # X-stabilizers (even parity sites)
        if (row + col) % 2 == 0:
            qubits = []
            # North neighbor
            if row > 0:
                qubits.append((row - 1) * d + col)
            # South neighbor
            if row < d - 1:
                qubits.append((row + 1) * d + col)
            # East neighbor
            if col < d - 1:
                qubits.append(row * d + (col + 1))
            # West neighbor
            if col > 0:
                qubits.append(row * d + (col - 1))

            if len(qubits) >= 2:  # Need at least 2-body for valid stabilizer
                x_stabilizers.append(qubits)
                x_stab_coords.append((row, col))

        # Z-stabilizers (odd parity sites)
        else:
            qubits = []
            # North neighbor
            if row > 0:
                qubits.append((row - 1) * d + col)
            # South neighbor
            if row < d - 1:
                qubits.append((row + 1) * d + col)
            # East neighbor
            if col < d - 1:
                qubits.append(row * d + (col + 1))
            # West neighbor
            if col > 0:
                qubits.append(row * d + (col - 1))

            if len(qubits) >= 2:
                z_stabilizers.append(qubits)
                z_stab_coords.append((row, col))
print(x_stabilizers)
print(x_stab_coords)
print(z_stabilizers)
print(z_stab_coords)
```
